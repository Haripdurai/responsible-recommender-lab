```python
from pathlib import Path
from urllib.request import urlopen
from urllib.error import URLError
import zipfile
import ssl

url = 'https://files.grouplens.org/datasets/movielens/ml-latest-small.zip'
zip_path = Path('ml-latest-small.zip')
dataset_dir = Path('ml-latest-small')

if not dataset_dir.exists():
    if not zip_path.exists():
        try:
            with urlopen(url) as response, open(zip_path, 'wb') as out_file:
                out_file.write(response.read())
        except URLError:
            # Fallback for environments with custom/self-signed cert chains.
            insecure_context = ssl._create_unverified_context()
            with urlopen(url, context=insecure_context) as response, open(zip_path, 'wb') as out_file:
                out_file.write(response.read())

    with zipfile.ZipFile(zip_path, 'r') as zf:
        zf.extractall('.')

print('Dataset ready:', dataset_dir.resolve())
```
